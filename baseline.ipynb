```python
import pandas as pd
```

```python
data = pd.read_csv("data.csv")
```

```python
data.head()
```

```python
data = data.sort_values("date").reset_index(drop = True)
```

```python
split_index = int(len(data) * 0.8)
train = data.iloc[:split_index]
test = data.iloc[split_index:]
print("train rows:" , len(train) , "test rows:" , len(test))
```

```python
volumes = data["volume_litres"].tolist()
```

```python
dates = []        # test ki dates
actuals = []      # asal volume
predictions = []  # baseline prediction
```

```python
for i in range(split_index, len(data)):
    last_7 = volumes[i-7: i]
    pred = sum(last_7) / 7
    dates.append(data["date"][i])
    actuals.append(volumes[i])
    predictions.append(round(pred, 2))
```

```python
total_abs_error = 0
total_pct_error = 0
```

```python
for k in range(len(actuals)):
    error = abs(actuals[k] - predictions[k])
    total_abs_error = total_abs_error + error
    total_pct_error = total_pct_error + (error/ actuals[k]) * 100
```

```python
mae = total_abs_error/len(actuals)
mape = total_pct_error/len(actuals)
print("MAE", round(mae,2))
print("MAPE", round(mape,2),"%")
```

```python
out = pd.DataFrame({
    "date": dates , "actual" : actuals , "baseline_pred" : predictions
})
out.to_csv("prediction.csv", index = False)
```

```python
report = "# Baseline Report: 7-day Moving Average\n\n"
report = report + "- Train rows: " + str(len(train)) + "\n"
report = report + "- Test rows: " + str(len(test)) + "\n"
report = report + "- MAE: " + str(round(mae, 2)) + " liters\n"
report = report + "- MAPE: " + str(round(mape, 2)) + " %\n"
with open("report.md", "w") as f:
    f.write(report)
```
